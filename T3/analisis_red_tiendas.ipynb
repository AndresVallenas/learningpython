```python
# 2. importamos librerias
import sys
import pandas as pd
import numpy as np
```

```python
# 3. cargar datos de lectura
# En la evaluacion, los ficheros CSV se encuentran en /workspace
# para no complicarnos, utilizo esta porcion del codigo para
# no realizar cambios manuales antes de entregar
from pathlib import Path

ruta_datos = Path("/workspace")

if not ruta_datos.exists():
    ruta_datos = Path(".")

df_sales = pd.read_csv(ruta_datos / "sales.csv")
df_inventories = pd.read_csv(ruta_datos / "inventories.csv")
df_satisfaction = pd.read_csv(ruta_datos / "satisfaction.csv")
```

```python
# Limpieza de datos 
df_sales = df_sales.dropna()
df_inventories = df_inventories.dropna()
df_satisfaction = df_satisfaction.dropna()

df_sales["Fecha_Venta"] = pd.to_datetime(df_sales["Fecha_Venta"])
df_inventories["Fecha_Actualización"] = pd.to_datetime(df_inventories["Fecha_Actualización"])
df_satisfaction["Fecha_Evaluación"] = pd.to_datetime(df_satisfaction["Fecha_Evaluación"])
```
